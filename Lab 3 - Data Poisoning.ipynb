```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification

# Create a small dataset
X, y = make_classification(
    n_samples=10,
    n_features=3,
    n_informative=2,
    n_redundant=0,
    random_state=42
)

# Dataset before poisoning
before = pd.DataFrame(
    X,
    columns=["Feature_1", "Feature_2", "Feature_3"]
)

before["Label"] = y

# Dataset after poisoning
after = before.copy()

# Select rows to poison
poison_rows = [1, 4, 7]

# Flip the labels
after.loc[poison_rows, "Label"] = (
    1 - after.loc[poison_rows, "Label"]
)

# Mark poisoned rows
before["Status"] = "Original"
after["Status"] = "Clean"

after.loc[poison_rows, "Status"] = "Poisoned"

# Display before and after side by side
side_by_side = pd.concat(
    {
        "BEFORE POISONING": before,
        "AFTER POISONING": after
    },
    axis=1
)

print("\nDATASET COMPARISON")
print(side_by_side.to_string())
```
